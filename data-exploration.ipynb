```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("dylanjcastillo/7k-books-with-metadata")

print("Path to dataset files:", path)
```

```python
import pandas as pd
```

```python
books = pd.read_csv(f'{path}/books.csv')
```

```python
books
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
ax = plt.axes()
sns.heatmap(books.isna().transpose(), cbar=False, ax=ax)
plt.xlabel("Columns")
plt.ylabel("Missing values")
plt.show()
```

```python
import numpy as np
books["missing_description"] = np.where(books["description"].isna(), 1, 0)
books["age_of_book"] = 2026 - books["published_year"]
```

```python
columns_of_interest = ["num_pages", "age_of_book", "missing_description", "average_rating"]
correlation_matrix = books[columns_of_interest].corr(method= "spearman")
sns.set_theme(style="white")
heatmap = sns.heatmap(correlation_matrix, annot=True, fmt=".2f", cmap="coolwarm", cbar_kws={"label": "Spearman correlation"})
heatmap.set_title("Correlation heatmap")
plt.show()
```

```python
books_clean = books.dropna(
    subset=["description", "num_pages", "average_rating", "published_year"]
)
```

```python
books_clean
```

```python
books_clean["categories"].value_counts().reset_index().sort_values("count", ascending=False)
```

```python
books_clean["words_in_description"] = books_clean["description"].str.split().str.len()
```

```python
books_clean
```

```python
books_clean.loc[books_clean["words_in_description"].between(1,4),"description"]
```

```python
books_clean.loc[books_clean["words_in_description"].between(5,14),"description"]
```

```python
books_clean.loc[books_clean["words_in_description"].between(25,34),"description"]
```

```python
books_clean_25_words = books_clean[books_clean["words_in_description"]>=25]
```

```python
books_clean_25_words
```

```python
books_clean_25_words["title_and_subtitle"] = np.where(
    books_clean_25_words["subtitle"].isna(),
    books_clean_25_words["title"],
    books_clean_25_words["title"] + ": " + books_clean_25_words["subtitle"]
)
```

```python
books_clean_25_words
```

```python
books_clean_25_words["tagged_description"] = books_clean_25_words[["isbn13","description"]].astype(str).agg(" ".join, axis=1)
```

```python
books_clean_25_words
```

```python
(
    books_clean_25_words
    .drop(["subtitle", "missing_description", "age_of_book", "words_in_description"], axis=1)
    .to_csv("books_cleaned.csv", index=False)
)
```
